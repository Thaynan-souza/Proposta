```python
import sqlite3
import pandas as pd

# Conectando ao banco que criamos
conexao = sqlite3.connect('../data/processed/banco_tcc.db')

# Consultando a tabela com os cabeçalhos renomeados
query = """
    SELECT codigo_setor, nome_bairro, total_domicilios, renda_media_domiciliar 
    FROM tb_censo_campinas_limpo 
    WHERE renda_media_domiciliar IS NOT NULL 
    LIMIT 10
"""

df_amostra = pd.read_sql(query, conexao)
conexao.close()

# Exibindo os dados formatados
display(df_amostra)
```
