```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Carregar os dados que acabamos de processar
caminho_csv = "../data/processed/esus_campinas_2023.csv"
df = pd.read_csv(caminho_csv, low_memory=False)

# 2. Visualizar as 5 primeiras linhas e entender as colunas
display(df.head())

# 3. Ver o total de linhas e colunas
print(f"Total de registros: {df.shape[0]}")
print(f"Total de colunas: {df.shape[1]}")

# 4. Criar um gráfico simples (Ex: Distribuição de Atendimentos por Sexo)
# A coluna 'AP_SEXO' está presente no padrão SIA/DATASUS
if 'AP_SEXO' in df.columns:
    df['AP_SEXO'].value_counts().plot(
        kind='bar', 
        color=['#1f77b4', '#ff7f0e'], 
        figsize=(8, 5)
    )
    plt.title('Atendimentos por Sexo - Campinas (2023)')
    plt.xlabel('Sexo')
    plt.ylabel('Quantidade de Atendimentos')
    plt.xticks(rotation=0)
    plt.show()
else:
    print("Coluna AP_SEXO não encontrada. Colunas disponíveis:", df.columns)
```

```python
# 1. Lista o nome exato de todas as 78 colunas para fácil visualização
print("LISTA DE TODAS AS COLUNAS:")
print(df.columns.tolist())

# 2. Mostra o tipo de dado de cada coluna e a quantidade de valores nulos (vazios)
print("\nESTRUTURA E DADOS VAZIOS:")
df.info(verbose=True, show_counts=True)

# 3. Gera um resumo estatístico rápido das colunas numéricas (médias, valores máximos, etc.)
display(df.describe())
```

```python
# Agrupando os dados pelo código da unidade de saúde (CNES)
df_entidades = df.groupby('AP_CODUNI').agg(
    Total_Atendimentos=('AP_CODUNI', 'count'),
    Valor_Total_Aprovado=('AP_VL_AP', 'sum')
).reset_index()

# Ordenando para ver quais unidades tiveram o maior volume de atendimentos
df_entidades = df_entidades.sort_values(by='Total_Atendimentos', ascending=False)

# Formatando a coluna de valor para exibir como moeda (R$)
df_entidades['Valor_Total_Aprovado'] = df_entidades['Valor_Total_Aprovado'].apply(lambda x: f"R$ {x:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.'))

print("Top 10 Entidades (CNES) com mais atendimentos:")
display(df_entidades.head(10))

# Criando um gráfico das 10 unidades com mais atendimentos
df_entidades.head(10).plot(
    x='AP_CODUNI', 
    y='Total_Atendimentos', 
    kind='bar', 
    color='#2ca02c', 
    figsize=(10, 5),
    legend=False
)
plt.title('Top 10 Unidades (CNES) em Volume de Atendimentos - Campinas')
plt.xlabel('Código CNES da Unidade')
plt.ylabel('Total de Atendimentos')
plt.xticks(rotation=45)
plt.show()
```

```python
# Verificando a distribuição por Raça/Cor na base
if 'AP_RACACOR' in df.columns:
    print("--- DISTRIBUIÇÃO POR RAÇA/COR ---")
    # Mapeamento padrão do DATASUS para Raça/Cor
    mapeamento_raca = {
        1: 'Branca',
        2: 'Preta',
        3: 'Parda',
        4: 'Amarela',
        5: 'Indígena',
        99: 'Sem informação'
    }
    df['Raca_Desc'] = df['AP_RACACOR'].map(mapeamento_raca).fillna('Não informado')
    display(df['Raca_Desc'].value_counts())
    print("\n" + "="*40 + "\n")

# Verificando a distribuição por Sexo
if 'AP_SEXO' in df.columns:
    print("--- DISTRIBUIÇÃO POR SEXO ---")
    display(df['AP_SEXO'].value_counts())
    
```

```python
# Verificando os principais CIDs informados (motivo/diagnóstico)
if 'AP_CID_CO' in df.columns:
    print("--- PRINCIPAIS CIDS REGISTRADOS (AP_CID_CO) ---")
    display(df['AP_CID_CO'].value_counts().head(10))
    print("\n" + "="*40 + "\n")

# Verificando os principais procedimentos realizados
if 'AP_PRIPAL' in df.columns:
    print("--- PRINCIPAIS PROCEDIMENTOS (AP_PRIPAL) ---")
    display(df['AP_PRIPAL'].value_counts().head(10))
    
```

```python
# Dicionário prático com o significado das principais CIDs encontradas
dicionario_cid = {
    'E039': 'Hipotireoidismo não especificado',
    'F329': 'Episódio depressivo não especificado',
    'J450': 'Asma predominantemente alérgica',
    'J45': 'Asma',
    'J459': 'Asma não especificada',
    'E03': 'Outros hipotireoidismos'
}

# Criando uma cópia para exibir com a tradução descritiva
df_cids_traduzido = df['AP_CID_CO'].value_counts().reset_index()
df_cids_traduzido.columns = ['CID', 'Quantidade']
df_cids_traduzido['Significado'] = df_cids_traduzido['CID'].map(dicionario_cid).fillna('Outra / Não mapeada neste lote')

display(df_cids_traduzido.head(10))
```

```python
# Verificando quantas linhas possuem CID principal preenchido vs vazias
total_linhas = len(df)
cids_preenchidos = df['AP_CID_CO'].notna().sum()
cids_vazios = df['AP_CID_CO'].isna().sum()

print(f"Total de registros na base: {total_linhas}")
print(f"Registros com CID preenchido (AP_CID_CO): {cids_preenchidos}")
print(f"Registros sem CID (vazios/NaN): {cids_vazios}")

# Vamos olhar também nas outras colunas de CID caso tenham dados lá
for col in ['CO_CIDPRIM', 'CO_CIDSEC', 'AP_CID_C1']:
    if col in df.columns:
        preenchidos_col = df[col].notna().sum()
        print(f"Registros com dados em {col}: {preenchidos_col}")
```

```python
# Verificando os 15 CIDs primários mais comuns na base de Campinas
print("--- TOP 15 CIDS PRIMÁRIOS MAIS FREQUENTES ---")
top_cids = df['CO_CIDPRIM'].value_counts().head(15).reset_index()
top_cids.columns = ['CID', 'Quantidade']

# Exibe a tabela bruta dos códigos
display(top_cids)
```

```python
# Verificando os procedimentos principais mais frequentes realizados em Campinas
print("--- TOP PROCEDIMENTOS PRINCIPAIS (AP_PRIPAL) ---")
top_procedimentos = df['AP_PRIPAL'].value_counts().head(10).reset_index()
top_procedimentos.columns = ['Codigo_Procedimento', 'Quantidade']

display(top_procedimentos)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Mapeamento de Raça/Cor do DATASUS para exibição amigável
mapeamento_raca = {
    1: 'Branca',
    2: 'Preta',
    3: 'Parda',
    4: 'Amarela',
    5: 'Indígena',
    99: 'Sem informação'
}
df['Raca_Desc'] = df['AP_RACACOR'].map(mapeamento_raca).fillna('Não informado')

# 2. Criando faixas etárias baseadas na coluna AP_NUIDADE (assumindo formato padrão do SIA)
# O padrão do DATASUS usa códigos nas dezenas para indicar dias/meses/anos, 
# mas AP_NUIDADE costuma trazer a idade numérica direta para anos (quando o código em AP_COIDADE é 03 ou similar).
# Vamos filtrar e criar uma visualização limpa por Sexo e Raça:

print("--- CRUZAMENTO: SEXO vs RAÇA/COR ---")
tabela_cruzada = pd.crosstab(df['AP_SEXO'], df['Raca_Desc'], margins=True)
display(tabela_cruzada)

# 3. Gráfico da distribuição por Raça/Cor
plt.figure(figsize=(8, 4))
df['Raca_Desc'].value_counts().plot(kind='bar', color='#4c72b0')
plt.title('Distribuição dos Atendimentos por Raça/Cor - Campinas (2023)')
plt.xlabel('Raça/Cor')
plt.ylabel('Total de Atendimentos')
plt.xticks(rotation=0)
plt.show()

# 4. Gráfico da distribuição por Faixa Etária (usando AP_NUIDADE)
if 'AP_NUIDADE' in df.columns:
    plt.figure(figsize=(10, 4))
    df['AP_NUIDADE'].plot(kind='hist', bins=20, color='#55a868', edgecolor='black')
    plt.title('Distribuição por Idade dos Pacientes Atendidos - Campinas')
    plt.xlabel('Idade (Anos)')
    plt.ylabel('Frequência')
    plt.show()
```
